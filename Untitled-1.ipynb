```python
from matplotlib import pyplot as plt
import numpy as np
x=np.arange(0,10,2)
y=x*10
f1={'family':'arial','color':'g','size':'18'}
f2={'family':'calibri','color':'r','size':'18'}
plt.xlabel('X- Values',fontdict=f2)
plt.ylabel('X Multiples 10',fontdict=f2)
plt.title("Multiples of 10",fontdict=f1)
plt.show()
```

```python
#Vertical
from matplotlib import pyplot as plt
X=[2023,2024,2025,2026]
Y=[2500,1700,1500,2200]
Z=['r','y','g','b']
print(X,Y)
plt.xlabel("Year")
plt.ylabel("Placements")
plt.title("Annual Placement Analysis")
plt.bar(X,Y,color=Z,width=0.5)
plt.show()
```

```python
#Horizontal
from matplotlib import pyplot as plt
X=['A','B','C','D']
Y=[10,20,30,40]
V=['g','k','y','r']
print(X,Y)
plt.xlabel("Year")
plt.ylabel("Placements")
plt.title("Annual Placement Analysis")
plt.barh(X,Y,color=V,height=0.7)
plt.show()
```

```python
from matplotlib import pyplot as plt
Year=[2023,2024,2025,2026]
Placements=[2000,1200,1000,2400]
cse=[40,20,29,50]
it=[50,40,34,70]
ece=[10,50,20,26]
ece_starts=[cse[i]+it[i] for i in range(len(cse))]#here this is beacause cse  and it added and next is ece 
plt.bar(Year,cse,color='r',width=0.4)
plt.bar(Year,it,bottom=cse,color='g')
plt.bar(Year,ece,bottom=ece_starts,color='b')
plt.show()
```

```python
from matplotlib import pyplot as plt
import numpy as np
year=[2023,2024,2025,2026]
cse=[23,45,67,28]
it=[30,12,23,21]
ece=[23,12,21,13]
w=0.2
plt.figure(figsize=(10,5))#here we set figure size mean graph size
cse_bar=np.arange(len(year))
print (cse_bar)
it_bar=[i+w for i in cse_bar]
print(it_bar)
ece_bar=[w+i for i in it_bar]
print(ece_bar)
plt.bar(cse_bar,cse,width=w,label="CSE")
plt.bar(it_bar,it,width=w,label="IT")
plt.bar(ece_bar,ece,width=w,label="ECE")
plt.xticks(cse_bar+w,year)
plt.title('Placemennt Analysis')
plt.xlabel("Y E A R")
plt.ylabel("PLCEMENTS")
plt.legend()
plt.show()
```

```python
#pie chart
#parameters are :labels,startangle,explode,shadow,color
#autopct means to write the percentage on ppiechat[% how many digits and after decimal howmany digit Float %%]
```

```python
from matplotlib import pyplot as plt
student_performance=["Excellent","Good","Average","Bad"]
student_values=[10,17,20,3]
plt.figure(figsize=(6,8))
plt.pie(student_values,labels=student_performance,startangle=0,explode=[0.1,0,0,0],shadow=True,autopct="%2.1f%%")
plt.legend(loc='lower right',facecolor='w',fancybox=True)
plt.title("Student Performance",color='r')
plt.show()
```
